# Analyse libre de vos résultats

Les figures standard sont produites par `python3 -m ubiq.analyze <exp>`. Ce carnet sert à **aller plus loin** : figures spécifiques à votre variante, lecture fine d'un run, analyse d'un cas d'échec.

```python
import sys, os, glob, csv, json
sys.path.insert(0, os.path.abspath('..'))
import matplotlib.pyplot as plt
from ubiq import metrics
EXP = 'baseline'          # <- nom de votre expérience
RES = os.path.abspath(os.path.join('..', 'results', EXP))
summary = metrics.read_csv(os.path.join(RES, 'summary.csv'))
summary[:3]
```

## 1. Lire un run en détail (rep 0 d'une variante)

```python
rd = sorted(glob.glob(os.path.join(RES, '*', '*', 'rep0')))[0]
tasks = metrics.read_csv(os.path.join(rd, 'tasks.csv'))
ev = metrics.read_csv(os.path.join(rd, 'events.csv'))
t = [float(r['t_gen']) for r in tasks if r['latency_ms']]
lat = [float(r['latency_ms']) for r in tasks if r['latency_ms']]
fig, ax = plt.subplots(figsize=(9, 3))
ax.scatter(t, lat, s=3, color='#2a78d6')
for e in ev:
    if e['type'] in ('ap_down', 'uav_launch', 'edge_down', 'migration_start'):
        ax.axvline(float(e['t']), color='#52514e', lw=0.8)
ax.set_xlabel('temps (s)'); ax.set_ylabel('latence (ms)'); ax.set_title(os.path.relpath(rd, RES))
plt.show()
```

## 2. Où échouent les tâches ?

```python
from collections import Counter
Counter((r['phase'], r['status']) for r in tasks if r['status'] != 'ok').most_common(10)
```
